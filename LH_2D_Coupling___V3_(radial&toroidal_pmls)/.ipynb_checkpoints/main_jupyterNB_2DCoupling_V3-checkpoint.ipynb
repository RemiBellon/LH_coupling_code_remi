```python
from sklearn.gaussian_process import GaussianProcessRegressor
from sklearn.gaussian_process.kernels import Matern, ConstantKernel as C
from sklearn.preprocessing import StandardScaler 

clean_df = df.query("Gamma_E < 0.005").copy()
X = clean_df[['S_real', 'S_imag', 'p_degree', 'L_pml_ratio']]
Y = np.log10(clean_df['Gamma_E'].values)

scaler = StandardScaler() # normalization of variable

X_scaled = scaler.fit_transform(X)

kernel = C(1.0) * Matern(length_scale=1.0, nu=2.5)
gpr = GaussianProcessRegressor(kernel=kernel, n_restarts_optimizer=5, alpha=1e-4)
gpr.fit(X_scaled[:2000], Y[:2000])
```

```python
# --- Training and validation --- 
clean_df =  df.query('Gamma_E < 0.9').copy() # reduce the dataset to consider only sufficiently "good" reflection coefficient
print(f'Tot points = {len(df)}, clean dataset valid points = {len(clean_df)}.')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score

print('--- Train & validate the model ---')
print('--- Training: ---')
pml_features  = ['S_real', 'S_imag', 'p_degree', 'L_pml_ratio', 'n_e', 'n_para', 'theta_B']
X_inputs = clean_df[pml_features].values
Y_outputs = np.log(clean_df['Gamma_E'].values) # GPR doesn't manage correctly severals order of magnitude values 

# Scale the inputs to get the same mathematic weight (for the model)
scaler = StandardScaler() # x_scaled = (x - mean) / sigma
X_scaled = scaler.fit_transform(X_inputs)
# train_size_var = round(3*len(clean_df))/(4*len(df)) # 3/4 of clean dataset points
train_size_var = 3000
# Train on 3/4 of clean dataset and the remaining values are used for testing the model 
X_train, X_test, Y_train, Y_test = train_test_split(X_inputs, Y_outputs, train_size=train_size_var, random_state=42)
print(f'X_train={len(X_train)}, X_test={len(X_test)}, Y_train={len(Y_train)}), Y_test={len(Y_test)}') 
print('The train and test split is done')
# Initialize the kernel (learn the data and generate the corresponding between data correlation function)
kernel = C(1.0) * Matern(length_scale=1.0, nu=2.5) # length_scale = approx size of attraction basin, nu = ???
# training the model
# alpha = noise to avoid correlation matrix diag divergence , n_restats_optimizer = starting from different location (set of 7D coord to find the global min, random_stat=random seed, normalize_y=True:scale the values in [0,1] 
gpr = GaussianProcessRegressor(kernel=kernel, alpha=1e-2, n_restarts_optimizer=5, random_state=42, normalize_y=True) 
print('The model is trained')
gpr.fit(X_train, Y_train)

print('--- Validation: ---')
# Based on trained model: compute & predict the X_test Gamma_E coeff
# the objective is to compare those prediction to the corresponding (test) real gamma_E values 

Y_pred_test = gpr.predict(X_test)
```

```python

print(f'Y_pred_test = {len(Y_pred_test)}')

r2 = r2_score(Y_test, Y_pred_test)
print(f'r2 = {r2:.4f}')
```
